## Introduction

Arabic Sign Language (ArSL) plays a crucial role in enabling communication for the deaf and hard-of-hearing community. Developing machine learning models that can accurately recognize hand gestures corresponding to Arabic letters is an essential step toward building inclusive AI systems.

In this notebook, we focus on collecting a high-quality dataset of Arabic sign language letter gestures using a large collection of real images. The collected data will later be used to train a deep learning model for instant Arabic letter recognition.

### Objectives of this notebook:

In this phase, we will:
- Preparing and preparing images for :

- Capturing Hand Landmarks with MediaPipe

In this section, we will use MediaPipe Hands to extract 21 hand landmarks (x, y, z) from our existing images. We will save these digital landmarks along with their labels (Arabic letters) in a CSV file.

These landmarks will later be used as structured inputs for our machine learning model.


This phase is critical to ensure that the model learns from diverse, well-labeled, and correctly formatted training data.

## 1- Import Libraries

```python
import os
import cv2 as cv 
import numpy as np
import pandas as pd
import csv
import mediapipe as mp 
import pandas as pd 
```

<hr style="border: 1px solid #ccc; margin: 30px 0;">

```python
DATASET_PATH = r"C:\Users\Mohamed Eprahim\Desktop\New folder (4)\Arabic Sign Language Dataset"



mp_hands = mp.solutions.hands
hands = mp_hands.Hands(static_image_mode=True, max_num_hands=2, min_detection_confidence=0.5)
mp_draw = mp.solutions.drawing_utils



output_file = "sign_language_data.csv"
failed_images_file = "failed_images.txt"


failed_images = []

with open(output_file, mode='w', newline='') as file:
    writer = csv.writer(file)
    headers = ["letter"] + [axis + str(i) for i in range(21) for axis in ["x", "y", "z"]]
    writer.writerow(headers)
    

    for letter in os.listdir(DATASET_PATH):
        # print(letter)
        letter_path = os.path.join(DATASET_PATH, letter)
        # print(letter_path)
        for img_name in os.listdir(letter_path):

            img_path = os.path.join(letter_path, img_name)

            img_pixels=cv.imread(img_path)


            if img_pixels is None:

                failed_images.append(img_name)

                continue

            img_rgb = cv.cvtColor(img_pixels, cv.COLOR_BGR2RGB)

            results = hands.process(img_rgb)
            if results.multi_hand_landmarks:
               for hand_landmarks in results.multi_hand_landmarks:
                    
                    landmarks = []

                    for lm in hand_landmarks.landmark:
                         
                         landmarks.extend([lm.x, lm.y, lm.z])

                    writer.writerow([letter] + landmarks)

            else:
                failed_images.append(img_name)

with open(failed_images_file, "w" , encoding="utf-8") as file:
    for img in failed_images:
        file.write(f"{img}\n")

    print("The data has been successfully saved to the CSV file")
    print(f"Total Number of images not read or processed : {len(failed_images)}  of the original images ( 7,884 ) . ")
```

```python


failed_images_2 = set()
images_processed = 0  

with open(txt_file, "r", encoding="utf-8") as file:
    images_names = set(line.strip() for line in file.readlines())

with open(output_file, mode='a', newline='') as file:
    writer = csv.writer(file)

    for img_name in images_names:
        found = False
        for letter in os.listdir(DATASET_PATH):
            letter_path = os.path.join(DATASET_PATH, letter)
            img_path = os.path.join(letter_path, img_name)

            if os.path.exists(img_path):
                found = True
                images_processed += 1  
                img_pixels = cv.imread(img_path)

                if img_pixels is None:
                    
                    failed_images_2.add(img_name)
                    break  

                
                img_gray = cv.cvtColor(img_pixels, cv.COLOR_BGR2GRAY)
                clahe = cv.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
                img_enhanced = clahe.apply(img_gray)
                img_rgb = cv.cvtColor(img_enhanced, cv.COLOR_GRAY2RGB)

                
                results = hands.process(img_rgb)

                if results.multi_hand_landmarks:
                    for hand_landmarks in results.multi_hand_landmarks:
                        landmarks = []
                        for lm in hand_landmarks.landmark:
                            landmarks.extend([lm.x, lm.y, lm.z])
                        writer.writerow([letter] + landmarks)  
                else:

                    failed_images_2.add(img_name)

                break  

        if not found:

            failed_images_2.add(img_name)


print(f"Total Number of images not read or processed : {len(failed_images_2)}  of the original images {len(images_names)} . ")
```
